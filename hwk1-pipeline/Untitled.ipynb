```python
import pandas as pd
```

```python
gt_filepath = 'https://d37ci6vzurychx.cloudfront.net/trip-data/green_tripdata_2025-11.parquet'

zones_filepath = 'https://github.com/DataTalksClub/nyc-tlc-data/releases/download/misc/taxi_zone_lookup.csv'

green_taxi_df = pd.read_parquet(
    path=gt_filepath
)

zones_dtypes= {
    "passenger_count": "Int64",
    "trip_distance": "float64",
    "RatecodeID": "Int64",
    "store_and_fwd_flag": "string",
    "PULocationID": "Int64",
    "DOLocationID": "Int64"}

zones= pd.read_csv(
    zones_filepath,
    dtype= zones_dtypes
    
)
```
